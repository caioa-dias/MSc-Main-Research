```python
import joblib
import pandas as pd
import numpy as np
from keras.models import load_model

data = pd.read_csv(r'C:\Users\caiod\Desktop\Mestrado - EESC-USP\Pesquisa Principal\Scripts\DataProcessing\Numerical-PressureDistributionData.csv', sep=',')

scaler_cl = joblib.load('scalers/scaler_cl.pkl')
scaler_cp = joblib.load('scalers/scaler_cp.pkl')
scaler_x = joblib.load('scalers/scaler_x.pkl')

input_cols = data.columns[0:3]
X = data[input_cols].values
X_scaled = scaler_x.transform(X)
```

```python
modelo = load_model('LF_numerical_ann.keras')

Y_pred_scaled = modelo.predict(X_scaled)
```

```python
Y_pred_cl_scaled = Y_pred_scaled[:, 0].reshape(-1, 1)
Y_pred_cp_scaled = Y_pred_scaled[:, 1:]

Y_pred_cl = scaler_cl.inverse_transform(Y_pred_cl_scaled)
Y_pred_cp = scaler_cp.inverse_transform(Y_pred_cp_scaled)

Y_pred = np.hstack([Y_pred_cl, Y_pred_cp])

united = np.hstack((X, Y_pred))

df = pd.DataFrame(united)

#df.to_csv('Numerical-PressureDistributionData-Predictions.csv', index=False)
colunas_x = ['Re', 'AoA', 'y']
colunas_y = ['cl'] + [f'cp-{i+1}' for i in range(Y_pred.shape[1] - 1)]

df.columns = colunas_x + colunas_y

df
```
